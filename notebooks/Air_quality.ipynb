##Dealing with Missing values

#Train Test Split

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=100)
```

```python
X_train.shape
```

##Standard Scaling

```python
from sklearn.preprocessing import StandardScaler
scalar=StandardScaler()
```

```python
X_train_scaled=scalar.fit_transform(X_train)
X_test_scaled=scalar.transform(X_test)
```

###Adding Polynomial features

```python
from sklearn.preprocessing import PolynomialFeatures
poly = PolynomialFeatures(degree=2, include_bias=False)
X_train_poly = poly.fit_transform(X_train_scaled)
X_test_poly  = poly.transform(X_test_scaled)
```

#Training the Model

```python
reg=LinearRegression()
lasso=LassoCV()
ridge=RidgeCV()
elastic=ElasticNetCV()
```

```python
reg.fit(X_train_scaled,y_train)
```

```python
y_pred=reg.predict(X_test_scaled)
plt.scatter(y_test,y_pred)
```

```python
mae=mean_absolute_error(y_test,y_pred)
score=r2_score(y_test,y_pred)
mse=mean_squared_error(y_test,y_pred)
rmse=np.sqrt(mse)
print("The mean absolute error for Linear reg model is:",mae)
print("The mean squared error for Linear reg model is:",mse)
print("The root mean squared error for Linear reg model is:",mae)
print("The r2 score is:",score)
```

###Based on Polynomial Features to Better Understand the Importance of Lasso

```python
from sklearn.linear_model import LinearRegression, LassoCV, RidgeCV, ElasticNetCV
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np


reg     = LinearRegression()
lasso   = LassoCV(cv=5, max_iter=10000)
ridge   = RidgeCV(cv=5)
elastic = ElasticNetCV(
    l1_ratio=[0.1, 0.3, 0.5, 0.7, 0.9, 0.95, 1.0],
    cv=5,
    max_iter=10000
)

models = {
    "Linear Regression": reg,
    "LassoCV":           lasso,
    "RidgeCV":           ridge,
    "ElasticNetCV":      elastic
}

results = {}

for name, model in models.items():
    model.fit(X_train_poly, y_train)
    y_pred_poly = model.predict(X_test_poly)

    mae  = mean_absolute_error(y_test, y_pred_poly)
    mse  = mean_squared_error(y_test, y_pred_poly)
    rmse = np.sqrt(mse)
    r2   = r2_score(y_test, y_pred_poly)

    results[name] = {"R2": round(r2, 4), "RMSE": round(rmse, 4), "MAE": round(mae, 4)}
    print(f"\n")
    print(f"Model: {name}")
    print(f"  R²   : {r2:.4f}")
    print(f"  RMSE : {rmse:.4f}")
    print(f"  MAE  : {mae:.4f}")


results_df = pd.DataFrame(results).T.sort_values("R2", ascending=False)
print("\n<---       Model Comparison       --->")
print(results_df)
```
